## 🔹 Paso 2: Analizar si el negocio es rentable

### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

Se usan los 3 datasets (`orders`, `catalog`, `marketing_spend`):

**📊 Parte 1: Rentabilidad del negocio**
- ¿Cuál es el ingreso total (revenue)? 
- ¿Cuál es el costo total? 
- ¿Cuánto se ha invertido en marketing? 
- ¿El negocio es rentable? (calcular profit)  

---

**📈 Parte 2: Comportamiento de ventas**
- ¿Cuál es el ticket promedio por orden? 
- ¿Cuál es la cantidad promedio de productos por orden? 
- ¿Cuál es el producto más vendido?
- ¿Cuánto se ha gastado en marketing por canal? 

```python
#PARTE 1: Rentabilidad del negocio
orders_catalog = orders.merge(catalog, on='nombre_producto', how='left') #Une los dataset impios
ingreso_total = orders_catalog['monto_total'].sum() #Ingreso total o Revenue
orders_catalog['costo_total_producto'] = orders_catalog['costo_unitario'] * orders_catalog['cantidad']
costo_total = orders_catalog['costo_total_producto'].sum() #Costo total de productos vendidos
inversion_marketing = marketing['gasto'].sum() #Inversion de markeing
beneficio_bruto= ingreso_total - costo_total#Beneficio bruto
beneficio_neto= beneficio_bruto - inversion_marketing#Beneficio neto

print(f"--- RENTABILIDAD DEL NEGOCIO ---")
print(f"Ingreso total (Revenue): ${ingreso_total:,.2f}")
print(f"Costo total de productos: ${costo_total:,.2f}")
print(f"Inversion total de marketing: ${inversion_marketing:,.2f}")
print(f"Beneficio bruto: ${beneficio_bruto:,.2f}")
print(f"Beneficio neto:${beneficio_neto:,.2f}")

```

```python

#PARTE 2:Comportamiento de ventas
ticket_promedio = orders_catalog['monto_total'].mean() #Ticket promedio
cantidad_promedio = orders_catalog['cantidad'].mean() #Cantidad promedio de producto
producto_mas_vendido = orders_catalog['nombre_producto'].value_counts().idxmax()#producto mas vendido
gasto_marketing_canal = marketing.groupby('canal')['gasto'].sum() #gasto de marketing por canal

print(f"--- COMPORTAMIENTO DE VENTAS ---")
print(f"Ticket Promedio: ${ticket_promedio:,.2f}")
print(f"Cantidad Promedio de Productos por Orden: {cantidad_promedio:,.2f}")
print(f"Producto Más Vendido:",producto_mas_vendido)
print(f"\nGasto en Marketing por Canal:\n{gasto_marketing_canal}")
```

```python
# Explorar tabla events
# =========================
query_events = '''
SELECT *
FROM events;
'''
events = pd.read_sql(query_events, con=engine)
events.head()
```

```python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''
SELECT 
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS usuarios_unicos
FROM events
GROUP BY nombre_evento
ORDER BY 
    CASE nombre_evento
        WHEN 'first_visit' THEN 1
        WHEN 'add_to_cart' THEN 2
        WHEN 'add_payment_info' THEN 3
        WHEN 'purchase' THEN 4
        ELSE 5
    END;
'''
totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = """
WITH cte_signup AS (
    SELECT DISTINCT id_usuario FROM events WHERE nombre_evento = 'first_visit'
), 
cte_cart AS (
    SELECT DISTINCT id_usuario FROM events WHERE nombre_evento = 'add_to_cart'
), 
cte_payment AS (
    SELECT DISTINCT id_usuario FROM events WHERE nombre_evento = 'add_payment_info'
), 
cte_purchase AS (
    SELECT DISTINCT id_usuario FROM events WHERE nombre_evento = 'purchase'
) 
SELECT 
    (SELECT COUNT(*) FROM cte_signup) AS signup_users,
    (SELECT COUNT(*) FROM cte_cart) AS cart_users,
    (SELECT COUNT(*) FROM cte_payment) AS payment_users,
    (SELECT COUNT(*) FROM cte_purchase) AS purchase_users,
    
    -- Drop-off entre signup y add_to_cart (%)
    ROUND(((SELECT COUNT(*) FROM cte_signup) - (SELECT COUNT(*) FROM cte_cart)) * 100.0 / NULLIF((SELECT COUNT(*) FROM cte_signup), 0), 2) AS dropoff_after_signup_pct,
    
    -- Drop-off entre add_to_cart y add_payment_info (%)
    ROUND(((SELECT COUNT(*) FROM cte_cart) - (SELECT COUNT(*) FROM cte_payment)) * 100.0 / NULLIF((SELECT COUNT(*) FROM cte_cart), 0), 2) AS dropoff_after_cart_pct,
    
    -- Drop-off entre add_payment_info y purchase (%)
    ROUND(((SELECT COUNT(*) FROM cte_payment) - (SELECT COUNT(*) FROM cte_purchase)) * 100.0 / NULLIF((SELECT COUNT(*) FROM cte_payment), 0), 2) AS dropoff_after_payment_pct;
"""

conversion = pd.read_sql(text(query_conversion), con=engine)
conversion
```

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users;
'''
users = pd.read_sql(query_users, con=engine)
users.head(3)
```

```python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT * 
FROM user_activity;
'''
user_activity = pd.read_sql(text(query_user_activity), con=engine)
user_activity.head(3)
```

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''
WITH user_cohorts AS (
    SELECT 
        id_usuario,
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte_mes
    FROM users
),
user_activity_months AS (
    SELECT DISTINCT
        uc.id_usuario,
        uc.cohorte_mes,
        (EXTRACT(YEAR FROM CAST(ua.fecha_actividad AS DATE)) - EXTRACT(YEAR FROM uc.cohorte_mes)) * 12 +
        (EXTRACT(MONTH FROM CAST(ua.fecha_actividad AS DATE)) - EXTRACT(MONTH FROM uc.cohorte_mes)) AS mes_numero
    FROM user_activity ua
    JOIN user_cohorts uc ON ua.id_usuario = uc.id_usuario
),
retencion AS (
    SELECT 
        uc.cohorte_mes,
        COUNT(DISTINCT uc.id_usuario) AS clientes_iniciales,
        COUNT(DISTINCT CASE WHEN uam.mes_numero = 1 THEN uam.id_usuario END) AS retained_m1,
        COUNT(DISTINCT CASE WHEN uam.mes_numero = 2 THEN uam.id_usuario END) AS retained_m2,
        COUNT(DISTINCT CASE WHEN uam.mes_numero = 3 THEN uam.id_usuario END) AS retained_m3
    FROM user_cohorts uc
    LEFT JOIN user_activity_months uam ON uc.id_usuario = uam.id_usuario
    GROUP BY uc.cohorte_mes
)
SELECT
    TO_CHAR(cohorte_mes, 'YYYY-MM') AS cohorte,
    clientes_iniciales,
    ROUND(retained_m1::numeric / NULLIF(clientes_iniciales, 0), 2) AS mes_1,
    ROUND(retained_m2::numeric / NULLIF(clientes_iniciales, 0), 2) AS mes_2,
    ROUND(retained_m3::numeric / NULLIF(clientes_iniciales, 0), 2) AS mes_3
FROM retencion
ORDER BY cohorte_mes;
'''

# Ejecutar la consulta
cohorte_final = pd.read_sql(text(query_cohort_retention_final), con=engine)
cohorte_final
```

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** No existe diferencia; ambas versiones del producto tienen el mismo impacto 
   - **H₁ (Hipótesis alternativa):** existe diferencia; las versiones del producto tienen impactos diferentes
   
**Test estadístico:** Prueba z para comparar proporciones  
**Nivel de significancia alpha:** .05

```python
from statsmodels.stats.proportion import proportions_ztest

experiment = pd.read_csv('https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv')
experiment.head()
```

```python
#Agrupamos y sumamos casos de exito
conversiones = experiment.groupby('variante')['convirtio'].sum()
observaciones = experiment.groupby('variante')['convirtio'].count()
exitos = [conversiones['control'], conversiones['tratamiento']]

#Obtenemos la tasa o porcentaje de exito
tasa_control = exitos[0] / observaciones[0]
tasa_tratamiento = exitos[1] / observaciones[1]
print(f"Tasa de conversión (Control): {tasa_control:.2%}")
print(f"Tasa de conversión (Tratamiento): {tasa_tratamiento:.2%}")

#Interpretacion de direccion de resultados
if tasa_tratamiento > tasa_control:
    print(f"\nLa variante de tratamiento tiene una mayor tasa de conversión ({tasa_tratamiento - tasa_control:.2%}).")
elif tasa_control > tasa_tratamiento:
    print(f"\nLa variante de control tiene una mayor tasa de conversión ({tasa_control - tasa_tratamiento:.2%})")
else:
    print("\nAmbas variantes tienen exactamente la misma tasa de conversión.")
```

```python
#Aplicamos la prueba Z de proporciones
z_stat, p_value = proportions_ztest(exitos, observaciones)

print(f"Estadístico Z: {z_stat:.4f}")
print(f"P-value: {p_value:.5f}")

#Interpretacion de resultados
alpha = 0.05
if p_value < alpha:
    print("Conclusión: Rechazamos la hipótesis nula (H0). Hay una diferencia estadísticamente significativa entre las variantes.")
else:
    print("Conclusión: No podemos rechazar la hipótesis nula (H0). No hay evidencia estadística suficiente para afirmar que las variantes difieren.")
```
